```python
# Combination of radiobutton and checkbutton
import tkinter as tk

def update_label():
    # Update the label text based on the selected options
    label.config(text=f"Selected: {var.get()}, Check: {check_var.get()}")

# Create the main window
root = tk.Tk()
root.title("Checkbutton and Radiobutton Example")

# Create a Label to display the selected option
label = tk.Label(root, text="Selected: Option 1, Check: Off")
label.pack()

# Create a StringVar to hold the Radiobutton selection
var = tk.StringVar()
var.set("Option 1")

# Create Radiobuttons
radio1 = tk.Radiobutton(root, text="Option 1", variable=var, value="Option 1", command=update_label)
radio2 = tk.Radiobutton(root, text="Option 2", variable=var, value="Option 2", command=update_label)
radio1.pack()
radio2.pack()

# Create a Checkbutton
check_var = tk.BooleanVar()
check_button = tk.Checkbutton(root, text="Check", variable=check_var, command=update_label)
check_button.pack()

# Create a button to quit the application
quit_button = tk.Button(root, text="Quit", command=root.destroy)
quit_button.pack()

# Run the main loop
root.mainloop()
```

```python
## Creating submenus
from tkinter import *

root = Tk()
menu = Menu(root)
root.config(menu=menu)

filemenu = Menu(menu)
menu.add_cascade(label='File', menu=filemenu)
filemenu.add_command(label='New')
filemenu.add_command(label='Open...')

# Create a submenu under "File" called "Recent"
recentmenu = Menu(filemenu)
filemenu.add_cascade(label='Recent', menu=recentmenu)
recentmenu.add_command(label='File 1')
recentmenu.add_command(label='File 2')

filemenu.add_separator()
filemenu.add_command(label='Exit', command=root.destroy)

helpmenu = Menu(menu)
menu.add_cascade(label='Help', menu=helpmenu)
helpmenu.add_command(label='About')

mainloop()
```

```python
from tkinter import *
root = Tk() 
scrollbar = Scrollbar(root) 
scrollbar.pack(side = RIGHT, fill = Y) 
mylist = Listbox(root, yscrollcommand = scrollbar.set) 
for line in range(100): 
    mylist.insert(END, 'This is line number' + str(line)) 
mylist.pack( side = LEFT ) 
scrollbar.config( command = mylist.yview ) 
mainloop() 

#The END argument specifies that each new line is added at the end of the list, and it includes a string with the line number.
#The Listbox is packed into the root window on the left side and set to fill both the horizontal and vertical space.
```

```python
import tkinter as tk
from tkinter import messagebox

root = tk.Tk()
root.title("Text Entry GUI")

# Create and place the text entry widget
text_entry = tk.Entry(root)
text_entry.pack(padx=20, pady=20)

def print_text():
    text = text_entry.get()
    print("Text entered:", text)

def quit_program():
    if messagebox.askyesno("Quit", "Do you want to quit?"):
        root.destroy()

# Create and place the "Ok" button
ok_button = tk.Button(root, text="Ok", command=print_text)
ok_button.pack(side=tk.LEFT, padx=5, pady=10)

# Create and place the "Quit" button
quit_button = tk.Button(root, text="Quit", command=quit_program)
quit_button.pack(side=tk.RIGHT, padx=5, pady=10)

root.mainloop()
```

```python
# Create a Scale widget for zooming
import tkinter as tk

def update_text_zoom(val):
    zoom_level = int(val)
    text.config(font=("Arial", zoom_level))

window = tk.Tk()
window.title("Text Zoom")

zoom_scale = tk.Scale(window, from_=10, to=30, orient="horizontal", label="Zoom Level", command=update_text_zoom)
zoom_scale.pack()

# Create a Text widget
text = tk.Text(window, wrap="word")
text.insert("1.0", "This is some example text that you can zoom in and out.")
text.pack()
window.mainloop()
# "1.0"--line.column," where "line" is the line number and "column" is the character position within that line.
```
