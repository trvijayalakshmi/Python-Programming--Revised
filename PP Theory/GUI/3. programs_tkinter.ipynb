```python
## Digital clock

import tkinter as tk
import time

# Create the GUI Window
root = tk.Tk()
root.title("Digital Clock")

# Create a Label for Displaying Time
time_label = tk.Label(root, font=("Helvetica", 48), bg="black", fg="white")
time_label.pack(fill="both", expand=True)

# Define a Function to Update Time
def update_time():
    current_time = time.strftime("%H:%M:%S")
    time_label.config(text=current_time)
    time_label.after(1000, update_time)  # Update every 1000ms (1 second)

# Run the Update Function and Tkinter Main Loop
update_time()  # Start updating the time
root.mainloop()  # Run the Tkinter main loop
```

```python
## Simple Animation pgm
from tkinter import *

# This program shows how to use the "after" function to make animation.

class Test(Frame):
    def createWidgets(self):
        self.QUIT = Button(self, text='QUIT', foreground='red',
                           command=self.destroy)
        self.QUIT.pack(side=LEFT, fill=BOTH)
        self.draw = Canvas(self, width="5i", height="5i")
# 5i is 5 inches
        self.draw.create_rectangle(0, 0, 10, 10, tags="thing", fill="blue")
        self.draw.pack(side=LEFT)

    def moveThing(self, *args):
        # move 1/10 of an inch every 1/10 sec (1" per second, smoothly)
        self.draw.move("thing", "0.01i", "0.01i")
        self.after(10, self.moveThing)

    def __init__(self, master=None):
        Frame.__init__(self, master)
        Pack.config(self)
        self.createWidgets()
        self.after(10, self.moveThing)

test = Test()
test.mainloop()
```

```python
from tkinter import *

# this uses subclassing. note that there is no explicit call to start Tk.
# Tkinter is smart enough to start the system if it's not already going.

class Test(Frame):
    def createWidgets(self):
        self.QUIT = Button(self, text='QUIT', foreground='red',
                           command=self.destroy)
        self.QUIT.pack(side=BOTTOM, fill=BOTH)
        self.draw = Canvas(self, width="5i", height="5i")
        self.speed = Scale(self, orient=HORIZONTAL, from_=-100, to=100)
        self.speed.pack(side=BOTTOM, fill=X)
        self.draw.create_rectangle(0, 0, 10, 10, tags="thing", fill="blue")
        self.draw.pack(side=LEFT)

    def moveThing(self, *args):
        velocity = self.speed.get()
        str = float(velocity) / 1000.0
        str = "%ri" % (str,)
        self.draw.move("thing",  str, str)
        self.after(10, self.moveThing)

    def __init__(self, master=None):
        Frame.__init__(self, master)
        Pack.config(self)
        self.createWidgets()
        self.after(10, self.moveThing)

test = Test()
test.mainloop()
```

```python
# Show how to do switchable panels.

from tkinter import *

class App:

    def __init__(self, top=None, master=None):
        if top is None:
            if master is None:
                top = Tk()
            else:
                top = Toplevel(master)
        self.top = top
        self.buttonframe = Frame(top)
        self.buttonframe.pack()
        self.panelframe = Frame(top,  borderwidth=2, relief=GROOVE)
        self.panelframe.pack(expand=1, fill=BOTH)
        self.panels = {}
        self.curpanel = None

    def addpanel(self, name, klass):
        button = Button(self.buttonframe, text=name,
                        command=lambda self=self, name=name: self.show(name))
        button.pack(side=LEFT)
        frame = Frame(self.panelframe)
        instance = klass(frame)
        self.panels[name] = (button, frame, instance)
        if self.curpanel is None:
            self.show(name)

    def show(self, name):
        (button, frame, instance) = self.panels[name]
        if self.curpanel:
            self.curpanel.pack_forget()
        self.curpanel = frame
        frame.pack(expand=1, fill="both")

class LabelPanel:
    def __init__(self, frame):
        self.label = Label(frame, text="Welcome to MGIT")
        self.label.pack()

class ButtonPanel:
    def __init__(self, frame):
        self.button = Button(frame, text="Press ON")
        self.button.pack()

def main():
    app = App()
    app.addpanel("ON", LabelPanel)
    app.addpanel("OFF", ButtonPanel)
    app.top.mainloop()

if __name__ == '__main__':
    main()
```

```python
""""Paint program by Dave Michell.

  All this does is put up a canvas and draw a smooth black line
whenever you have the mouse button down, but hopefully it will
be enough to start with.. It would be easy enough to add some
options like other shapes or colors...
"""

from tkinter import *

"""paint.py: not exactly a paint program.. just a smooth line drawing demo."""

b1 = "up"
xold, yold = None, None

def main():
    root = Tk()
    drawing_area = Canvas(root)
    drawing_area.pack()
    drawing_area.bind("<Motion>", motion)
    drawing_area.bind("<ButtonPress-1>", b1down)
    drawing_area.bind("<ButtonRelease-1>", b1up)
    root.mainloop()

def b1down(event):
    global b1
    b1 = "down"           # you only want to draw when the button is down
                          # because "Motion" events happen -all the time-
def b1up(event):
    global b1, xold, yold
    b1 = "up"
    xold = None           # reset the line when you let go of the button
    yold = None

def motion(event):
    if b1 == "down":
        global xold, yold
        if xold is not None and yold is not None:
            event.widget.create_line(xold,yold,event.x,event.y,smooth=TRUE)
                          # here's where you draw it. smooth. neat.
        xold = event.x
        yold = event.y

if __name__ == "__main__":
    main()
```

```python
import tkinter as tk
from tkinter import ttk
import sounddevice as sd
import soundfile as sf
import numpy as np
import threading
import datetime
import os

class VoiceRecorder:
    def __init__(self, root):
        self.root = root
        self.root.title("Voice Recorder")

        self.record_button = ttk.Button(root, text="Record", command=self.start_recording)
        self.stop_button = ttk.Button(root, text="Stop", command=self.stop_recording, state="disabled")
        self.volume_slider = ttk.Scale(root, from_=0, to=100, orient="horizontal", label="Volume")
        self.status_label = ttk.Label(root, text="Status: Idle")

        self.record_button.pack(pady=10)
        self.stop_button.pack(pady=10)
        self.volume_slider.set(50)
        self.volume_slider.pack(pady=10)
        self.status_label.pack()

        self.recording = False
        self.thread = None
        self.audio_data = None
        self.silence_threshold = 0.03  # Adjust this value based on your environment

    def start_recording(self):
        self.recording = True
        self.record_button["state"] = "disabled"
        self.stop_button["state"] = "enabled"
        self.status_label["text"] = "Status: Recording"

        self.thread = threading.Thread(target=self.record_audio)
        self.thread.start()

    def stop_recording(self):
        self.recording = False

    def record_audio(self):
        duration = 5  # Recording duration in seconds
        sample_rate = 44100  # Sample rate (samples per second)

        self.audio_data = []

        while self.recording:
            audio_chunk, overflowed = sd.rec(int(sample_rate), samplerate=sample_rate, channels=2, dtype=np.int16)
            audio_chunk = (audio_chunk * (self.volume_slider.get() / 100.0)).astype(np.int16)
            self.audio_data.extend(audio_chunk)

            if self.detect_silence(audio_chunk, self.silence_threshold):
                self.recording = False

        sd.stop()
        self.save_audio()

    def detect_silence(self, audio_chunk, threshold):
        # Check if the audio_chunk contains silence (volume below the threshold)
        rms = np.sqrt(np.mean(audio_chunk ** 2))
        return rms < threshold

    def save_audio(self):
        if self.audio_data:
            filename = f"recording_{datetime.datetime.now().strftime('%Y-%m-%d_%H-%M-%S')}.wav"
            sf.write(filename, self.audio_data, 44100)
            self.status_label["text"] = f"Recording saved as {filename}"
        else:
            self.status_label["text"] = "Recording failed."

        self.record_button["state"] = "enabled"
        self.stop_button["state"] = "disabled"

if __name__ == "__main__":
    root = tk.Tk()
    app = VoiceRecorder(root)
    root.mainloop()
```

```python
## File system GUI
import os
import tkinter as tk
from tkinter import filedialog, messagebox

# Function to open a file dialog for directory selection
def open_directory():
    directory_path = filedialog.askdirectory()
    if directory_path:
        load_directory(directory_path)

# Function to load and display files and directories in the selected directory
def load_directory(directory_path):
    current_directory.set(directory_path)
    file_listbox.delete(0, tk.END)
    for item in os.listdir(directory_path):
        file_listbox.insert(tk.END, item)

# Function to open the selected file or directory
def open_item():
    selected_item = file_listbox.get(file_listbox.curselection())
    path = os.path.join(current_directory.get(), selected_item)
    if os.path.isfile(path):
        messagebox.showinfo("File Selected", f"Selected file: {selected_item}")
    else:
        load_directory(path)

# Create the main application window
app = tk.Tk()
app.title("File System GUI")

# Create a label to display the current directory
current_directory = tk.StringVar()
directory_label = tk.Label(app, textvariable=current_directory)

# Create a listbox to display files and directories
file_listbox = tk.Listbox(app, selectmode=tk.SINGLE, height=15, width=50)

# Create buttons to open a directory and open a selected item
open_directory_button = tk.Button(app, text="Open Directory", command=open_directory)
open_item_button = tk.Button(app, text="Open Item", command=open_item)

# Pack the widgets into the window
directory_label.pack(pady=10)
file_listbox.pack(pady=10)
open_directory_button.pack()
open_item_button.pack()

# Load the current directory
load_directory(os.getcwd())

# Start the main loop
app.mainloop()
```

```python
## A button to retrieve the next value in that list(if there 
# is one).This button is displayed if there is no next value
# in the list.
# A label to display the number of the items being displayed
# and the total number of items
import tkinter as tk

class ItemDisplayApp:
    def __init__(self, master):
        self.master = master
        self.master.title("Item Display App")

        self.item_list = ["Item 1", "Item 2", "Item 3", "Item 4", "Item 5"]
        self.current_item_index = 0

        self.label = tk.Label(master, text="", font=("Helvetica", 16))
        self.label.pack(pady=10)

        self.show_next_button = tk.Button(master, text="Show Next", command=self.show_next_item)
        self.show_next_button.pack()

        self.update_label()

    def show_next_item(self):
        if self.current_item_index < len(self.item_list):
            self.current_item_index += 1
            self.update_label()

    def update_label(self):
        if self.current_item_index < len(self.item_list):
            current_item = self.item_list[self.current_item_index]
            self.label.config(text=f"Item: {current_item}  (Item {self.current_item_index + 1}/{len(self.item_list)})")
            self.show_next_button.config(state="normal")
        else:
            self.label.config(text="All items displayed")
            self.show_next_button.config(state="disabled")

if __name__ == "__main__":
    root = tk.Tk()
    app = ItemDisplayApp(root)
    root.mainloop()
```
