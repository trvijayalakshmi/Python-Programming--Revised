```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x300")  # Set initial window size

label1 = tk.Label(root, text="Label 1", bg="red", fg="white")
label1.place(x=50, y=50)

label2 = tk.Label(root, text="Label 2", bg="green", fg="white")
label2.place(x=150, y=100)

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x200")  # Set initial window size

label1 = tk.Label(root, text="Label 1", bg="red", fg="white")
label1.place(relx=0.1, rely=0.1)  # Place at 10% of width and 10% of height

label2 = tk.Label(root, text="Label 2", bg="green", fg="white")
label2.place(relx=0.5, rely=0.5)  # Place at 50% of width and 50% of height

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x200")  # Set initial window size

label1 = tk.Label(root, text="Label 1", bg="red", fg="white")
label1.place(x=50, y=50, anchor=tk.NE)  # Place using North-East anchor

label2 = tk.Label(root, text="Label 2", bg="green", fg="white")
label2.place(relx=0.5, rely=0.5, anchor=tk.CENTER)  # Place at center using Center anchor

root.mainloop()
```

```python
import tkinter as tk

root = tk.Tk()
root.geometry("300x300")
label1 = tk.Label(root, text="Name:", bg="red", fg="white")
label1.grid(row=0, column=0, sticky=tk.W, padx=5, pady=5)

entry1 = tk.Entry(root)
entry1.grid(row=0, column=1, padx=5, pady=5)

label2 = tk.Label(root, text="Email:", bg="green", fg="white")
label2.grid(row=1, column=0, sticky=tk.W, padx=5, pady=5)

entry2 = tk.Entry(root)
entry2.grid(row=1, column=1, padx=5, pady=5)

button1 = tk.Button(root, text="Submit")
button1.grid(row=2, column=0, columnspan=2, pady=10)

root.mainloop()
```
