## 1. Encapsulation

```python
Encapsulation means wrapping data (attributes) and methods 
(functions) into a single unit (class).

It restricts direct access to some variables and methods, 
protects the internal state of an object.

Achieved in Python using access modifiers.
```

### Access Modifiers in Python

Unlike languages like Java or C++, Python does not have strict access control keywords. Instead, it uses convention and name mangling:

```python
1. Public members (default)-> Accessible from anywhere.
         Example: self.name

2. Protected members (_single underscore) 
   A convention: should not be accessed directly outside the class.
         Example: self._salary

3. Private members (__double underscore)
   Name mangled → can’t be accessed directly outside the class.
         Example: self.__password
```

```python
Why Encapsulation?

✅ Data protection (prevent accidental changes).
✅ Controlled access (via getters/setters).
✅ Flexibility (internal logic can be changed without affecting 
               external code).
✅ Code maintenance and readability.
```

```python
class Student:
    def __init__(self, name, marks, grade):
        self.name = name            # Public
        self._marks = marks         # Protected
        self.__grade = grade          # Private

    def display(self):
        print(f"Name: {self.name}, Marks: {self._marks}, Grade: {self.__grade}")

s = Student("Anu", 90, 'A')
s.display()

print(s.name)       # ✅ Public accessible
print(s._marks)     # ⚠️ Accessible but not recommended
print(s.__grade)  # ❌ Error: private
```

```python
To Access Private Members (Name Mangling)
```

```python
print(s._Student__grade)   # ✅ Accessing private variable
```

```python
Simple Example of protected and private variables:
```

```python
class Student:
    def __init__(self, marks, grade):
        self._marks = marks  # protected
        self.__grade = grade # private

s = Student(95, 'A+')
print(s._marks)   # to access protected variable
print(s._Student__grade)    # to access private variable
```

```python
Thereby, in inheritance, encapsulation controls access to data.  
```

```python
Another Example of Encapsulation:
```

```python
class Bank:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    def deposit(self, amount):
        self.__balance += amount
    
    def withdraw(self, amount):
        self.__balance -= amount

    def show_balance(self):
        print("Balance:", self.__balance)
    
    def check_balance(self):
        if self.__balance < 1000:
            print("Minimum balance not maintained")
        else:
            print("Balance is sufficient")

b = Bank(1000)
b.deposit(500)
b.withdraw(200)
b.show_balance()
b.check_balance()

# print(b.__balance)
print(b._Bank__balance)
```

```python
Eg of Private Methods
```

```python
class Bank:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance      # Private attribute (data hidden)

    # Public method — accessible from outside
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ₹{amount}. New balance: ₹{self.__balance}")

    # Public method — accessible from outside
    def withdraw(self, amount):
        if self.__validate_withdrawal(amount):   # calls private method internally
            self.__balance -= amount
            print(f"Withdrew ₹{amount}. Remaining: ₹{self.__balance}")
        else:
            print("Insufficient funds!")

    # Private method — hidden from outside, used internally only
    def __validate_withdrawal(self, amount):
        return amount <= self.__balance

    # Public method to safely read balance
    def get_balance(self):
        return self.__balance

acc = Bank("Aisha", 5000)

acc.deposit(1000)         # ✅ Works — public method
acc.withdraw(2000)        # ✅ Works — public method
print(acc.get_balance())  # ✅ Works — public method

print(acc._Bank__validate_withdrawal(100))  #accessing private method
# acc.__validate_withdrawal(100)  # ❌ Error — private method, can't call directly
# print(acc.__balance)            # ❌ Error — private attribute, hidden
```

## 2. Polymorphism:

```python
It allows different classes to define methods with the same name
but different behaviors.

"One name, many forms."

Same method name behaves differently depending on context.
```

```python
Eg of Function polymorphism: (Same function with different behavior)
```

```python
print(len("abcd"))
print(len([1,2,3]))
```

```python
Why Polymorphism?

✅ Improves code flexibility and reusability
✅ Makes code easier to maintain and extend
✅ Reduces duplication of logic
```

```python
Polymorphism using a common method or function
```

```python
class Manager:
    def salary(self):
        print("Manager salary: 80000")

class Developer:
    def salary(self):
        print("Developer salary: 60000")

class Intern:
    def salary(self):
        print("Intern stipend: 15000")

Manager().salary()
Developer().salary()
Intern().salary()
```

### Polymorphism overcomes:

```python
1. Method Overriding: A subclass provides a method already defined
   in its parent class.
2. Method Overloading: Same method name with different parameters.
```

1. Method Overriding:

If the same method is present in both the superclass (parent) and subclass (child)

In this case, the method in the subclass overrides the method in the superclass. This concept is known as method overriding in Python.

```python
Ex: Overriding method
```

```python
class Parent:    # Superclass
    def display(self):
        print("Display from Parent")

class Child(Parent):   # Subclass
    def display(self):    # Overriding
        print("Display from Child")

c = Child()  # Object creation
c.display()  
```

```python
How to overcome Overriding:
Solution: Python achieves polymorphism by using super() method or 
          by calling Parent's method directly to overcome 
          “Method Overriding” 
    
Solution 1. Use super() to call parent’s method inside child method
Solution 2. Call Parent’s method directly (avoiding super)
```

```python
To overcome "Method Overriding" (example of Single inheritance)
```

```python
class Person:   # Parent class
    def __init__(self, name):
        self.name = name
    def display(self):
        print("Employee Name:", self.name)

class Employee(Person):   # Inheriting from Employee (Child class)
    def display(self):    
        print("Role: Teamleader")

m = Employee("Ravi")
m.display()     # calling overrided method
```

```python
class Person:   # Parent class
    def __init__(self, name):
        self.name = name
    def display(self):
        print("Employee Name:", self.name)

class Employee(Person):   # Inheriting from Employee (Child class)
    def display(self):
        Person.display(self)
#         super().display()
        print("Role: Teamleader")

m = Employee("Ravi")
m.display()     # calling overrided method
```

```python
In inheritance, polymorphism is achieved by overriding methods.
```

```python
Ex2: To overcome "Method Overriding" (Hierarchical inheritance)
```

```python
class Shape:
    def area(self):
        print("To calculate Area of a Circle or Square")

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
#         super().area()
        Shape.area(self)
        print("area of a circle is", 3.14 * self.r * self.r)        
        
class Square(Shape):
    def __init__(self, s):
        self.s = s
    def area(self):
#         super().area()
        Shape.area(self)
        print("area of a square is", self.s * self.s)        
        
c = Circle(5)   # Object creation
c.area()

d = Square(4)   # Object creation
d.area()
```

### 2. Method Overloading:
    
Python does not support method overloading directly (same method name with different parameters).

```python
How python achieves polymorphism without "Method Overloading"?

Solution: Python achieves polymorphism without "Method Overloading" 
          by using default arguments or *args.
```

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

c = Calculator()
print(c.add(2, 3))       # 5
print(c.add(2, 3, 4))    # 9
```

```python
class Calculator:
    def add(self, *args):
        return sum(args)

c = Calculator()
print(c.add(2, 3))       # 5
print(c.add(2, 3, 4))    # 9
print(c.add(5, 7, 9, 12)) 
```

### Operator Overloading (Special case in python): 

Same operator behaves differently depending on the operands.
Thereby Python achieves polymorphism without "Overloading operator"

```python
Eg:1 The same operator can be used for both bitwise operations
and logical operations, depending on the context.
```

```python
# Bitwise AND
print(5 & 3)  # Output: 1 (binary: 0101 & 0011)

# Logical AND
print(True & False)  # Output: False
```

```python
Eg:2 Operators might perform differently based on the data types
they are working with.
```

```python
# Addition for integers
print(1 + 2)  

# Concatenation for strings
print("Hello" + " " + "World")  # Output: Hello World

x = [2,4,5]
y = [5,3,2]
print(x+y)  # concatenates two lists
```
