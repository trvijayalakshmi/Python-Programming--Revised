## Python Inheritance: 

```python
Inheritance allows a class (child) to inherit properties and methods
from another class (parent).
```

```python
Uses of Inheritance:

a) Since a child class can inherit all the functionalities of the
   parent's class, this allows code reusability.
b) Once a functionality is developed, you can simply inherit it. 
   No need to define them again.
c) Since you can also add your own functionalities in the child 
   class, you can inherit only the useful functionalities and define 
   other required features.
```

```python
Types of Inheritance:
    1. Single inheritance
    2. Multiple Inheritance
    3. Multilevel Inheritance
    4. Hierarchical Inheritance
```

```python
1. Single Inheritance

A child class inherits from one parent class.
```

```python
class Parent():
    def show_p(self):
        print("I am Parent")

class Child(Parent):
    def show_c(self):
        print("I am Child")

c = Child()  # Object creation with child class
c.show_c()
c.show_p()
```

```python
class Person:  # Parent class
    def basic_info(self):
        print("Learning things to find a job in a company")

class Employee(Person):   # Child class
    def role(self):
        print("Attend works assigned by the company")

# Create object of child class
m = Employee()

m.basic_info()     # inherited method
m.role()   # child class method
```

```python
2. Multilevel Inheritance

A class is derived from another derived class (chain).
```

```python
class Grandparent:
    def show_gp(self):
        print("I am Grandparent")

class Parent(Grandparent):
    def show_p(self):
        print("I am Parent")

class Child(Parent):
    def show_c(self):
        print("I am Child")

c = Child()  # Object creation with child class
c.show_gp()
c.show_p()
c.show_c()
```

```python
Simple Example of Multilevel inheritance where Person → Employee → Manager.
```

```python
class Person:
    def basic_info(self):
        print("Learning things to find a job in a company")

class Employee(Person):
    def role(self):
        print("Attend works assigned by the company")

class Manager(Employee):
    def decision(self):
        print("Manager makes decisions")

# Create object
s = Manager()

s.basic_info()   # from Person
s.role()         # from Employee
s.decision()     # from Manager or own method
```

```python
Complex Example of Multilevel Inheritance where Person → Employee → Manager.
```

```python
class Person:
    def __init__(self, name): 
        self.name = name

class Employee(Person):  # Employee (inherits from Person)
    def __init__(self, name, emp_id):
        Person.__init__(self, name)
        self.emp_id = emp_id  # additional attribute: emp_id
    def employee_details(self):
        print(f"Employee name is {self.name} with ID {self.emp_id}")

class Manager(Employee):   # Manager (inherits from Employee)
    def __init__(self, name, emp_id, dept):
        Employee.__init__(self, name, emp_id)
        self.dept = dept  # additional attribute: dept.

    def display(self):
        print(f"Name: {self.name}, ID: {self.emp_id}, Dept: {self.dept}")

m = Manager("Rajesh", 101, "ECE")
m.display()
m.employee_details()
```

```python
3. Multiple Inheritance

A child class inherits from more than one parent class.
```

```python
class Father:
    def skill1(self):
        print("Father: Gardening")

class Mother:
    def skill2(self):
        print("Mother: Cooking")

class Child(Father, Mother):   # Inheriting from both
    def skill3(self):
        print("Child: Playing")

# Create object
c = Child()

c.skill1()   # from Father
c.skill2()   # from Mother
c.skill3()   # own method
```

```python
Simple example of Multiple Inheritance:
```

```python
class Teacher:
    def teach(self):
        print("Teaches the subject Python")

class Researcher:
    def research(self):
        print("Does research in Deep learning")
    
class Professor(Teacher, Researcher):   # Inherites both the features
    def display(self):
        print("Teaches the subjects and does research")

p = Professor()
p.teach()
p.research()
p.display()
```

```python
Complex Example of Multiple Inheritance:
```

```python
# Parent Class 1
class Teacher:
    def __init__(self, name, subject):
        self.name = name
        self.subject = subject

    def teach(self):
        print(f"{self.name} is teaching {self.subject}.")

# Parent Class 2, 
class Researcher:
    def __init__(self, field):
        self.field = field

    def research(self):
        print(f"She is researching in the field of {self.field}.")

# Derived Class (inherits from both Teacher and Researcher)
class Professor(Teacher, Researcher):
    def __init__(self, name, subject, field):
        Teacher.__init__(self, name, subject)      # Initialize Teacher part
        Researcher.__init__(self, field)  # Initialize Researcher part

    def display(self):
        print(f"Prof: {self.name}, Subject: {self.subject}, Field: {self.field}")

# Object creation
p = Professor("ECE", "Python Programming", "Deep learning")
p.display()
p.teach()
p.research()
```

```python
4. Hierarchical Inheritance

Multiple child classes inherit from the same parent class.
```

```python
Simple example of hierarchical inheritance
```

```python
# Parent class
class Employee:
    def display_company(self):
        print("Company: TCS")

# Child class 1
class Manager(Employee):
    def show_role(self):
        print("Role: Manager")

# Child class 2
class Developer(Employee):
    def show_role(self):
        print("Role: Developer")

# Creating objects
m = Manager()
d = Developer()

# Accessing methods
m.display_company()
m.show_role()

d.display_company()
d.show_role()
```

```python
Example 2 of Hierarchical inheritance
```

```python
class Shape:
    pi = 3.14   # class variable (common for all shapes)
    def __init__(self, length, breadth, radius):
        self.length = length      # instance variable
        self.breadth = breadth    # instance variable
        self.radius = radius      # instance variable

class Rectangle(Shape):
    def area(self):
        return self.length * self.breadth
    
    def perimeter(self):
        return 2 * (self.length + self.breadth)

class Circle(Shape):
    def area(self):
        return Shape.pi * self.radius * self.radius
    
    def perimeter(self):
        return 2 * Shape.pi * self.radius

# Creating objects
r = Rectangle(5,3,4)
c = Circle(5,3,4)

print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())

print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())
```

```python
Complex Example of Hierarchical Inheritance:
```

```python
# Parent class
class Polygon:
    def __init__(self, n):
        self.n = n
        self.sides = [0 for i in range(n)]

    def inputSides(self):
        self.sides = [float(input("Enter side "+str(i+1)+" : ")) for i in range(self.n)]

    def dispSides(self):
        for i in range(self.n):
            print("Side",i+1,"is",self.sides[i])
            
poly = Polygon(4)    
poly.inputSides()
poly.dispSides()
```

```python
A Rectangle is a polygon with 4 sides. So, we can create a class 
called Rectangle which inherits from Polygon. This makes all the 
attributes and methods of Polygon class available to the Rectangle
class.

We don't need to define them again (code reusability). Rectangle 
can be defined as follows.
```

```python
class Rectangle(Polygon):
    def __init__(self):
        Polygon.__init__(self,2)

    def findAreaPeri(self):
        a, b = self.sides
        s = 2*(a + b) 
        print('The perimeter of the rectangle is %0.2f' %s)
        area = a * b
        print('The area of the rectangle is %0.2f' %area)
        
# Creating an instance of the Rectangle class
t = Rectangle()
t.inputSides()
t.dispSides()
# Calculating and printing the area of the rectangle
t.findAreaPeri()
```

```python
You can create child 2 class "Triangle"
```

```python
# Child 1 class
class Triangle(Polygon):
    def __init__(self):
        Polygon.__init__(self,3)

    def findAreaPeri(self):
        a, b, c = self.sides
        s = (a + b + c) 
        print('The perimeter of the triangle is %0.2f' %s)
        area = (s*(s-a)*(s-b)*(s-c)) ** 0.5
        print('The area of the triangle is %0.2f' %area)
        
# Creating an instance of the Triangle class
t = Triangle()
t.inputSides()
t.dispSides()
# Calculating and printing the area of the triangle
t.findAreaPeri()
```
