## 1. Data Processing Script

```python
A small data processor that:

 - Reads a list of numbers
 - Cleans the data (removes invalid entries)
 - Performs analysis (mean, max, min, sum)
 - Provides results
```

```python
class DataProcessor:
    def __init__(self, data):
        self.data = data
    def clean_data(self):
        self.data = [x for x in self.data if type(x) in (int, float)]
    def summary(self):
        total = sum(self.data)
        avg = total / len(self.data)
        return total, avg
    
data = [10, 20, "abc", 30, None, 40]
obj = DataProcessor(data)
obj.clean_data()
print("Cleaned Data:", obj.data)
print("Summary:", obj.summary())
```

## 2. Basic Calculator

```python
class Calculator:
    def add(self, a, b): return a + b
    def sub(self, a, b): return a - b
    def mul(self, a, b): return a * b
    def div(self, a, b): return a / b if b != 0 else "Error: Division by zero"

calc = Calculator()
print(calc.add(10, 5))
print(calc.sub(10, 5))
print(calc.mul(10, 5))
print(calc.div(10, 0))
```

```python
class Calculator:
    def __init__(self):
        print("Calculator Ready!")
    def add(self, a, b):
        return a + b
    def sub(self, a, b):
        return a - b
    def mul(self, a, b):
        return a * b
    def div(self, a, b):
        if b == 0:
            return "Error! Division by zero."
        return a / b
    def power(self, a, b):
        return a ** b

calc = Calculator()   # Object creation
print("Addition (10 + 5):", calc.add(10, 5))
print("Subtraction (10 - 5):", calc.sub(10, 5))
print("Multiplication (10 * 5):", calc.mul(10, 5))
print("Division (10 / 5):", calc.div(10, 5))
print("Power (2^3):", calc.power(2, 3))
```

## 3. File Organizer

```python
import os
import shutil

class FileOrganizer:
    def __init__(self, path):
        self.path = path

    def organize(self):
        for file in os.listdir(self.path):
            if file.endswith(".jpg") or file.endswith(".png"):
                folder = "Images"
            elif file.endswith(".pdf") or file.endswith(".txt"):
                folder = "Documents"
            else:
                folder = "Others"

            folder_path = os.path.join(self.path, folder)

            if not os.path.exists(folder_path):
                os.mkdir(folder_path)

            shutil.move(os.path.join(self.path, file), folder_path)

path = input("Enter folder path: ")
obj = FileOrganizer(path)
obj.organize()
print("Files Organized!")
```

```python
Elaborated version of File organizer
```

```python
import os
import shutil

class FileOrganizer:
    def __init__(self, directory):
        self.directory = directory

        # File type categories
        self.file_types = {
            "Images": [".jpg", ".jpeg", ".png", ".gif"],
            "Documents": [".pdf", ".docx", ".txt", ".pptx"],
            "Videos": [".mp4", ".mkv", ".avi"],
            "Music": [".mp3", ".wav"],
            "Programs": [".py", ".java", ".c", ".cpp"]
        }

    def create_folder(self, folder_name):
        path = os.path.join(self.directory, folder_name)
        if not os.path.exists(path):
            os.makedirs(path)

    def move_file(self, file, folder_name):
        source = os.path.join(self.directory, file)
        destination = os.path.join(self.directory, folder_name, file)
        shutil.move(source, destination)

    def organize_files(self):
        for file in os.listdir(self.directory):
            file_path = os.path.join(self.directory, file)

            # Skip directories
            if os.path.isdir(file_path):
                continue

            moved = False
            for folder, extensions in self.file_types.items():
                if file.lower().endswith(tuple(extensions)):
                    self.create_folder(folder)
                    self.move_file(file, folder)
                    moved = True
                    break

            # If file type not matched
            if not moved:
                self.create_folder("Others")
                self.move_file(file, "Others")

folder_path = input("Enter folder path to organize: ")
organizer = FileOrganizer(folder_path)
organizer.organize_files()
print("Files organized successfully!")
```

## 4. Simple Data Analysis with Pandas

```python
import pandas as pd

# Sample dataset (dictionary → DataFrame)
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 35, 40, 28],
    "Score": [85, 90, 88, 76, 95]
}

df = pd.DataFrame(data)

# 1. Display first few rows
print("Data:\n", df.head())

# 2. Basic statistics
print("\nSummary Statistics:\n", df.describe())

# 3. Average age and score
print("\nAverage Age:", df["Age"].mean())
print("Average Score:", df["Score"].mean())

# 4. Find highest score
topper = df.loc[df["Score"].idxmax()]
print("\nTop Scorer:\n", topper)

# 5. Filter students with score > 85
print("\nStudents with Score > 85:\n", df[df["Score"] > 85])
```
