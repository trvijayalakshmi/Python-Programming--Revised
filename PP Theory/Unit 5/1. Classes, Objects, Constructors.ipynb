## 1. Python Classes and Objects

```python
Python works with 

1. Functional programming
2. Procedure oriented programing (Functions within functions)
3. Modular programming
4. Object oriented programming 
```

```python
a) Class: A class is a blueprint for creating objects. It defines attributes 
   (variables) and methods (functions).
b) Object: An object is an instance of a class. Every object will have attributes 
   and methods of a class.
c) Attributes: Attributes are variables inside a class. They describe properties of
   an object.
d) Methods: Methods are functions inside a class. They define behavior of objects.
```

```python
Example of class, object, attributes, methods

The self parameter represents the current instance of the class.
```

```python
class Calculator:   # Class
    def add(self, a, b):   # Methods: add, sub
        return a + b
    def sub(self, a, b):    # Attributes: a, b
        return a - b

c = Calculator()     # Object creation with no parameters
print(c.add(10, 5))  # calling method with parameters
print(c.sub(10, 5))
```

```python
class Calculator:   # Class
    def set_values(self, a, b):  # Attributes: a, b
        self.a = a
        self.b = b
    def add(self):   # Methods: add, sub
        return self.a + self.b
    def sub(self):    
        return self.a - self.b

c = Calculator()     # Object creation with no parameters
c.set_values(10,5)  # calling method with parameters
print(c.add())     # calling method with no parameters
print(c.sub())
```

## 2. Constructor (__init__):   (The _ _init_ _() Function)


```python
A constructor is a special method in Python classes.
It is automatically called when an object is created.
  __init__ -> automatic initialization
    
__init__() constructor used to initialize object attributes when an 
object is created.
```

### Types of Constructors : 

```python
Parameterized constructor: Constructor with parameters is known as parameterized 
constructor. The parameterized constructor takes its first argument as a reference 
to the instance being constructed known as self and the rest of the arguments are 
provided by the programmer.

Default constructor: The default constructor is a simple constructor which doesn’t 
accept any arguments. Its definition has only one argument which is (self) reference to 
the instance being constructed.
```

```python
Eg of Parametrized Constructor:
```

```python
class Calculator:   # Class
    def __init__(self, a, b):   # Parametrized constructor
        self.a = a
        self.b = b
    def add(self):   # Methods: add, sub
        return self.a + self.b
    def sub(self):    # Attributes: a, b
        return self.a - self.b

c1 = Calculator(10,5)     # Object1 creation with parameters
print(c1.add())   # calling method with no parameters
print(c1.sub())

c2 = Calculator(100,40)  # Object2 creation
print(c2.add())
print(c2.sub())
```

```python
Define a class Rectangle with methods to compute area
and perimeter for it.
```

```python
class Rectangle:
    def __init__(self, a, b):   # ? constructor
        self.a = a
        self.b = b
    
    def area(self):
        return self.a * self.b
    def peri(self):
        return 2*(self.a + self.b)

r = Rectangle(10, 20)   # values passed at object creation
print("Area:", r.area())
print("Perimeter:", r.peri())
```

### 3. Class Variables and Instance Variables:

```python
class Student:
    college = "MGIT"  # Class variable
    
    def __init__(self, name, roll):  # parametrized constructor
        self.name = name      # instance variables or attributes
        self.roll = roll    
        
    def display(self):
        print(self.name, self.roll, Student.college)
        
s1 = Student("Ravi", 69)   # Object1 creation with parameters
s2 = Student("Sanjay", 85) # Object2 creation with parameters

s1.display()    # calling method with no parameters
s2.display()

print(s1.name, s1.roll)      # Accessing instance attributes or variables
print(s2.name, s2.roll)
print(s1.college)           # Accessing class attribute or variable
```

## 4. The self Parameter:

```python
The self parameter is a reference to the current instance of the class, and is used 
to access variables that belongs to the class.

It does not have to be named self, you can call it whatever you like, but it has 
to be the first parameter of any function in the class:
```

```python
#Use the words mysillyobject and abc instead of self:
class Student:
    def __init__(mysillyobject, name, roll):
        mysillyobject.name = name
        mysillyobject.roll = roll

    def greet(abc):
        print(abc.name, abc.roll)

# Creating objects (instances) of the class "Person"
s1 = Student("Sam", 301)

s1.greet()     # Calling method
```

## 5. Destructor:

```python
# Python program to illustrate destructor
class students:
 
    # default constructor
    def __init__(self):
        self.name = "ECE students"
    
    # a method for printing data members
    def display(self):
        print(self.name)
        
# creating object of the class
obj = students()
del obj      # Destructor

obj.display()    # try to use the method
```

## 6. Programs on Classes:

```python
1. Define a class "Book" with attributes "title, author, and price".
Write methods to display the book details and check if the book is 
expensive (price > 500).
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display(self):
        print("Title:", self.title)
        print("Author:", self.author)
        print("Price:", self.price)

    def is_expensive(self):
        if self.price > 500:
            print("The book is expensive")
        else:
            print("The book is affordable")

b1 = Book("Python Basics", "Guido", 650)
b1.display()
b1.is_expensive()
```

```python
2. Define a class with appropriate attributes and methods 
   to check whether a student has cleared the exam or not
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    def result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

s1 = Student("Anu", 35)
s1.display()
s1.result()
```

```python
3. Define a class with appropriate attributes and methods 
   for banking application.
```

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance  

    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        self.balance -= amount

    def show_balance(self):
        print("Balance:", self.balance)
    
    def check_balance(self):
        if self.balance < 1000:
            print("Minimum balance not maintained")
        else:
            print("Balance is sufficient")

b = Bank(1000)
b.deposit(500)
b.withdraw(200)
b.show_balance()
b.check_balance()
```

```python
4. Define a class with appropriate attributes and methods 
   to check the bonus eligibility of an employee.
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

    def bonus(self):
        if self.salary > 30000:
            print("Eligible for bonus")
        else:
            print("Not eligible for bonus")

e1 = Employee("Raj", 40000)
e1.display()
e1.bonus()
```
