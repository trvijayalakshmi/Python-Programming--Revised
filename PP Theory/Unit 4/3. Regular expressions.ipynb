# 1. Regular Expressions in Python

```python
A regular expression in a programming language is a special text
string used for describing a search pattern. 

It is extremely useful for extracting information from text such
as code, files, log, spreadsheets or even documents.

The most common uses of regular expressions are:

> Search a string (search and match)
> Finding a string (findall)
> Break string into a sub strings (split)
> Replace part of a string (sub)
```

### RegEx Module(re module) in Python

```python
Python has a built-in package called re Module, which can be 
used to work with Regular Expressions.

To use it, we need to import the module.

>import re
```

```python
Core functions of re module:

-->match: Returns a Match object if there is a match else 
          returns None.
-->search: Returns a Match object if there is a match anywhere 
           in the string.
-->findall: Returns a list containing all matches.
-->split: Returns a list where the string has been split at  
          each match.
-->sub: Replaces one or many matches with a string
```

```python
Useful Methods of re module:

group() → Returns the matched string.
start() → Start index of match.
end() → End index of match.
span() → Tuple of (start, end).
```

# 1. match()

```python
This method finds match if it occurs at start of the string. For
example, calling match() on the string ‘python programming’ and 
looking for a pattern ‘python’ will match. However, if we look for
only programming, the pattern will not match.
```

```python
Syntax:
re.match(pattern, string)
```

```python
import re
# matching python in the given sentence
result = re.match('python', 'python programming and python')
print(result)
print(result.group())
```

```python
import re
# matching programming in the given sentence
result = re.match('programming', 'python programming and python')
print(result)
```

```python
We have Member functions to this object to get specific data like
start, end and span
```

```python
import re
result = re.match('python', 'python programming and python')
print(result)
print('Starting position of the match:',result.start())
print('Ending position of the match :',result.end())
print('Span of the match :',result.span())
```

# 2. search()

```python
It is similar to match() but it doesn’t restrict us to find matches
at the beginning of the string only. The search() function searches
the string for a match, and returns a Match object if there is a
match.
```

```python
Syntax:
re.search(pattern, string)
```

```python
search() method is able to find a pattern from any position of the
string but it only returns the first occurrence of the search pattern.
```

```python
import re
result = re.search('python', 'python programming and python')
print(result.group())
result = re.search('programming', 'python programming and python')
print(result.group())
```

# 3. findall()

```python
The re.findall() method returns a list of all matches of a pattern
within the string. It returns the patterns in the order they are 
found. If there are no matches, then an empty list is returned.
```

```python
Syntax:  
re.findall(pattern, string)
```

```python
import re
ms = re.findall('python', 'python programming and python scripting')
print(ms)    
print(len(ms))
```

# 4. split()

```python
The re.split method splits the string where there is a match and 
returns a list of strings where the splits have occurred.
```

```python
Syntax:
re.split(pattern, string, [maxsplit=0])
```

```python
\s   It returns a match if the string contains any white space 
character.
```

```python
import re
# \s matches white spaces
resultls = re.split('\s', 'python programming and python scripting')
print(resultls)
#split with maxsplit
resultls = re.split('\s', 'python programming and python scripting',2)
print(resultls)
```

# 5. sub()

```python
The sub() function replaces the matches with the specified characters.
i.e., It helps to search a pattern and replace with a new sub 
string. If the pattern is not found, string is returned unchanged.
```

```python
Syntax:
re.sub(pattern, replace, string)
```

```python
import re
s='python programming and python scripting'
result = re.sub('python','java',s)
print(result)

str='python programming and python scripting'
#\s replace white space with hyphen
result = re.sub('\s','-',s)
print(result)
```

## 2. Programs on Regular expressions

```python
1. Write a python code to read a phone number and email-id from the
user and validate it for correctness.
```

```python
import re
# Regular expressions for validation
phone_pattern = r'\b[6-9]\d{9}\b' 
email_pattern = r'\b[\w\.-]+@[\w\.-]+\.\w{2,}\b'

# Read inputs
phone = input("Enter phone number: ")
email = input("Enter email ID: ")

# Validate phone
if re.match(phone_pattern, phone):
    print("Valid Phone Number")
else:
    print("Invalid Phone Number")

# Validate email
if re.match(email_pattern, email):
    print("Valid Email ID")
else:
    print("Invalid Email ID")
```

```python
# For phone no 
\b: This is a boundary anchor.
[6-9] Matches exactly one digit between 6 and 9.
\d{9} \d → any digit (0–9)  {9} → exactly 9 digits.
So this part matches 9 digits after the first digit.
```

```python
# For email
\w → letters, digits, underscore (a-z, A-Z, 0-9, _)
[] → character set
[\w\.-]+ (Username part)
. → dot
- → hyphen
+ → one or more characters
```

```python
2. Program to extract Phone Numbers
```

```python
import re

text = "Please call 67 555-123-4567 for assistance or 987-654-3210 for inquiries."

# Define a regular expression pattern for matching phone numbers
pattern = r'\b\d{3}-\d{3}-\d{4}\b'

# Use re.findall() to find all phone numbers in the text
phone_numbers = re.findall(pattern, text)

# Print the matched phone numbers
for phone in phone_numbers:
    print(phone)
```

```python
\b: This is a word boundary anchor. It matches the position 
where a word (sequence of word characters) starts or ends.

\w: This is a shorthand character class that matches word characters (letters, digits, and underscores).

\w+: This part matches the top-level domain (TLD) of the email address (e.g., com, org).

.-: They match a dot or a hyphen.

\.: This is a literal dot character.

[ ]: Square brackets denote a character class, which matches any single character that is contained within the brackets.
```

```python
3. Program to find the Email Addresses
```

```python
import re

text = "Please contact support@example.com for assistance or info@example.org for information."

# Define a regular expression pattern for matching email addresses
pattern = r'\b[\w.-]+@[\w.-]+\.\w+\b'

# Use re.findall() to find all email addresses in the text
email_addresses = re.findall(pattern, text)

# Print the matched email addresses
for email in email_addresses:
    print(email)
```

```python
4. Program to find the URLs
```

```python
import re

text = "Visit our website at http://www.example.com or check https://www.example.org for more information."

# Define a regular expression pattern for matching URLs
pattern = r'https?://\w+\.\w+\.\w+'

# Use re.findall() to find all URLs in the text
urls = re.findall(pattern, text)

# Print the matched URLs
for url in urls:
    print(url)
```

### 3. Special Symbols and MetaCharacters
A regular expression can be formed by using the mix of meta-characters and special sequences.

```python
Metacharacters are characters that are interpreted in a special way by a RegEx engine.

Metacharacter is a character with the specified meaning.
```

```python
Metacharacter	Description
------------    -----------
1. []---->         It represents the set of characters.

2. \ --->          It represents the special sequence.

3. ^ --->          It represents characters at the beginning
                of the string.
            
4. + --->          It represents one or more occurrences 
                of a pattern in the string.            
                
5. * --->          It represents zero or more occurrences 
                of a pattern in the string. 

6. ? --->          It represents zero or one occurrence
                of a pattern in the string.

7. $ --->          It represents characters at the end 
                of the string.

8. { } --->        It represents exactly the specified number
                of occurrences.

9. ( )--->         Capture and group(i.e group sub-patterns)                                                         
10. | --->          It represents either this or that character is 
                present.              
```

### 1. [ ] - Square brackets

```python
Square brackets specify a set of characters you wish to match.

You can also specify a range of characters using - inside square brackets.

[a-e] is the same as [abcde]
[1-4] is the same as [1234]
[0-39] is the same as [01239]
```

### 2. \ - Backslash

```python
Backlash \ is used to escape various characters including all metacharacters.

For example,
\s --> space

\d	It returns a match if the string contains digits [0-9].

\t --> tab space.

\b  This is a word boundary anchor.
```

### 3. ^ - Caret

```python
The caret symbol ^ is used to complement (invert) the character set. By using caret ^ symbol at the start of a square-bracket.

[^abc] means any character except a or b or c.
[^0-9] means any non-digit character.
```

### 4. + - Plus

```python
The plus symbol + matches one or more occurrences of the pattern left to it.

Expression      String     Matched
----------      ------     -------
ma+n	          mn	   No match (no a character)
	              man	   Match
	              maaan	   Match
	              main	   No match (a is not followed by n)
	              woman	   Match
```

### 5. * - Star


```python
The star symbol * matches zero or more occurrences of the pattern left to it.

Expression    String      Matched
---------     ------      -------
ma*n	        mn	       Match
	            man	       Match
	            maaan	   Match
	            main	   No match (a is not followed by n)
	            woman	   Match
```

```python
Matches zero or one occurrence of the preceding character or group.
For example, colou?r would match both "color" and "colour".
```

### 6. ? - Question mark

```python
Matches zero or one occurrence of the preceding character or group.
For example, colou?r would match both "color" and "colour".
```

### 7. $ - Dollar

```python
The dollar symbol $ is used to check if a string ends with a certain character.

Expression     String      Matched
-----------   --------     --------  
a$	            a		    match
	         formula		match
	           cab		    No match
```

### 8. { } - Braces

```python
{n,m}: This means at least n, and at most m repetitions of the pattern left to it.

Expression     String       Matched
---------      ------       -------
a{2,3}	       abc dat	    No match
	           abc daat	    1 match (at daat)
	           aabc daaat	2 matches (at aabc and daaat)
	           aabc daaaat	2 matches(at aabc and daaaat)
```

### 9. ( ) - Group

```python
Parentheses () is used to group sub-patterns. For example, (a|b|c)xz match any string that matches either a or b or c followed by xz

Expression      String       Matched
----------      ------       -------
(a|b|c)xz	    ab xz		 No match
		        abxz		 1 match (match at abxz)
		        axz cabxz	 2 matches (at axzbc cabxz)
```

### 10. | - Alternation

```python
Vertical bar | is used for alternation (or operator).

Expression     String      Matched
----------     ------      -------
a|b	            cde	       No match
	            ade	       1 match (match at ade)
	           acdbea	   3 matches (at acdbea)

Here, a|b match any string that contains either a or b             
```

```python
import re
string = 'Madhu:5231,Naveen:5832,Ganesh:509,Ramesh:569'
pattern = ':\d{3}'
result = re.split(pattern, string) 
print(result)

## The pattern :\d{2} in a regular expression is used to match
## a colon followed by exactly two digits (0-9).
```
