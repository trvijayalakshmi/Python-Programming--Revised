```python
File Handling in Python:

Python provides built-in functions to create, read, write, and
append files using the open() function.
```

```python
A file operation takes place in the following order:

1. Open a file
2. Read or write (perform operation)
3. Close the file
```

### 1. File Modes:

```python
Mode	Description
"x" 	Creates a new file, error if file exists.
"r" 	Read (default). Opens file, error if file doesn’t exist.
"w" 	Write. Creates new file or overwrites existing file.
"a" 	Append. Adds data at the end of the file.
"r+"	Read and Write.
"w+"	Write and Read (overwrites file).
"a+"	Append and Read.
```

```python
file = open("filename", "mode")
```

```python
2. Creating a file
```

```python
3. Writing to a File and Reading from a File
```

```python
f = open("sample.txt", "w")  # Overwrites if file exists
f.write("Hello, this is Python File Handling!\n")
f.write("Writing second line.\n")
f.close()

# Reading back
f = open("sample.txt", "r")
print(f.read())
f.close()
```

```python
4. Reading the whole File - read()

readline(): Reading line by line
```

```python
f = open("sample.txt", "r")
print(f.readline())    # Reads one line
f.close()
```

```python
readlines(): Reading all lines in a list
```

```python
f = open("sample.txt", "r")
print(f.readlines())   # Reads all lines into a list
f.close()
```

```python
writelines(): Writes multiple lines from a list
```

```python
f = open("sample.txt", "w")  # Overwrites if file exists
f.writelines(["Python is interesting. ", "But students are not showing interest"])
f.close()

# Reading back
f = open("sample.txt", "r")
print(f.read())
f.close()
```

```python
5. Appending to a File in append mode
```

```python
f = open("sample.txt", "a")  # Adds to existing content
f.write("This line is appended.\n")
f.close()

# Reading back
f = open("sample.txt", "r")
print(f.read())
f.close()
```

### 2. File Methods:

```python
Once a file is opened, you can use various methods to interact with it.

Method	Description
f.read()	Reads the file content (all or size bytes).
f.readline()	Reads a single line.
f.readlines()	Reads all lines into a list.
f.write(string)	Writes a string into the file.
f.writelines(list)	Writes a list of strings into the file.
f.close()	Closes the file.

f.tell()	Returns the current position of file pointer.
f.seek(offset)	Moves the file pointer to a specific position.
```

```python
tell() Method: It tells (returns) the current position of the file 
pointer in the file.
```

```python
f = open("sample.txt", "w") # Overwrites if file exists
f.write("Hello Python. This is Unit IV\n")
f.write("Hello Python. This is Unit III\n")
f.write("Hello Python. This is Unit II\n")
f.write("Hello Python. This is Unit I\n")
f.close()

f = open("sample.txt", "r")
print(f.read(5))      # Read first 5 characters
print("Current position:", f.tell())  # File pointer position
f.close()
```

```python
seek(offset) Method: Moves the file pointer to a specific
position.

offset: number of bytes to move.
```

```python
f = open("sample.txt", "r")
print(f.read(5))       # Read 5 characters → "Hello"
print(f.tell())
f.seek(9)              # Move pointer to beginning
print(f.read())        # Read full file again
f.close()
```

```python
f = open("sample.txt", "r")
f.seek(6)              # Move to position 6
print(f.read())        # Reads from position 6 → "Python"
f.close()
```

```python
Combination of tell() and seek()
```

```python
f = open("sample.txt", "r")
print("Initial position:", f.tell())

f.read(5)
print("After reading 5 chars:", f.tell())

f.seek(0)
print("After resetting:", f.tell())

f.close()
```

```python
Opening a file using with Statement (Best Practice)

It automatically closes the file after use.
```

```python
with open("sample.txt", "a") as f:
    f.write("Appended using WITH statement.\n")
    
with open("sample.txt", "r") as f:
    print(f.read())
```

### 3. Working with CSV files

```python
CSV (Comma-Separated Values) is a popular format for storing 
tabular data.
```

```python
1. Writing to a CSV file
```

```python
import csv
with open("students.csv", "w") as f:
    wr = csv.writer(f)
    wr.writerow(["Name", "Age", "Marks"])   # Header
    wr.writerow(["Archana", 20, 85])
    wr.writerow(["Raj", 22, 90])
    wr.writerow(["Charitha", 21, 78])
```

```python
2. Reading from a CSV file
```

```python
import csv
with open("students.csv", "r") as f:
    re = csv.reader(f)
    for row in re:
        print(row)
```

```python
import csv
with open("students.csv", "r") as f:
    re = csv.DictReader(f)
    for row in re:
        print(row)
```

### 4. Working with JSON Files

```python
JSON (JavaScript Object Notation) is used for storing structured
data (like dictionaries).
```

```python
a) CSV is best for tabular data (rows/columns).
b) JSON is best for structured/nested data.
```

```python
1. Writing to a JSON file
```

```python
import json

data = [
    {'Name': 'Archana', 'Age': '20', 'Marks': '85'},
    {'Name': 'Raj', 'Age': '22', 'Marks': '90'},
    {'Name': 'Charitha', 'Age': '21', 'Marks': '78'}
]

with open("student.json", "w") as f:
    json.dump(data, f) 
```

```python
2. Reading from a JSON file
```

```python
import json

with open("student.json", "r") as f:
    data = json.load(f)

print(data)
```

```python
json.dump() writes JSON data directly to a file.
json.dumps() converts a Python object to a JSON string.
```

### 5. Programs on Files:

```python
1. Write a Python code to merge two given file contents into a
third file.
```

```python
file1 = open("sample1.txt",'w')
file1.write("This is python programming lab")

file2 = open('sample2.txt','w') 
file2.write("\nThis content is not copyrighted")

file1 = open("sample1.txt",'r')
file2 = open("sample2.txt",'r')
output_file = open("merged.txt",'w')

for line in file1:
    output_file.write(line)
for line in file2:
    output_file.write(line)

output_file = open("merged.txt",'r')
print(output_file.read())
print("Files merged successfully")
```

```python
2. Write a program that reads a file and displays the number of
words, number of vowels, blank spaces, lower case letters and
uppercase letters.
```

```python
file = open('sample1.txt','r')
num_vowels = 0
num_spaces = 0
num_lower = 0
num_upper = 0

for line in file:
    words = line.split()
    for char in line:
        if char.lower() in 'aeiou':
            num_vowels = num_vowels + 1
        if char == ' ':
            num_spaces = num_spaces + 1
        if char.islower():
            num_lower = num_lower + 1
        if char.isupper():
            num_upper = num_upper + 1
        
print("Number of words:", len(words))
print(f"Number of vowels: {num_vowels}")   
print(f"Number of spaces: {num_spaces}")
print(f"Number of lowercase letters: {num_lower}")
print(f"Number of uppercase letters: {num_upper}")
```

```python
3. Write a Python code to open a given file and to check for given
words present in it and display on found.
```

```python
file = open('sample1.txt','r') 
words_to_search = ["write", "lab", "Python"]

content = file.read().lower()  # Convert to lowercase for case-insensitive search            
found_words = []
for i in words_to_search:
    if i.lower() in content:
        print(f"Found: {i}")
        found_words.append(i)
    else:
        print(f"Not Found: {i}")
```

```python
4. Write a Python code to Read text from a text file, find the
word with most number of occurrences
```

```python
file = open('merged.txt','r') 
word_counts = {}
common_word = None
max_occ = 0

for line in file:
    words = line.split()
    for word in words:
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1

        if word_counts[word] > max_occ:
            max_occ = word_counts[word]
            common_word = word
        
print(f"The most common word is '{common_word}' with {max_occ} occurrences.")
```
