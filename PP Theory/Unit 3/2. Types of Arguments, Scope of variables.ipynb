### 1. Types of Arguments

```python
Python supports different kinds of arguments:
a) Positional arguments → matched by order.
b) Keyword arguments → matched by key.
c) Default arguments - Setting a default value.
d) Variable length Positional arguments (*args)
e) Variable length Keyword arguments (*kargs)
```

```python
a) Positional arguments (order matters)
```

```python
def div(a,b):
    return a/b

print(div(8,2))
```

```python
Positional arguments with default value
```

```python
def div(a,b=4):
    return a/b

print(div(8,2))
print(div(8))
```

```python
Calculate compound interest given principal, rate and time with
defalut values
```

```python
def cal_interest(principal, rate=5, time=3):
    amount = principal * (1 + rate/100) ** time
    interest = amount - principal
    return interest

interest = cal_interest(30000)
print("The compound interest is:", interest)
interest = cal_interest(30000, 10)
print("The compound interest is:", interest)
interest = cal_interest(30000, 10, 4)
print("The compound interest is:", interest)
```

```python
b) Keyword arguments (name=value): Order doesn't matter
```

```python
def div(a,b):
    return a/b

print(div(a=8,b=2))
print(div(b=2,a=8))
```

```python
Keyword arguments with default value
```

```python
def div(a,b=4):
    return a/b

print(div(b=2,a=8))
print(div(a=8))
```

```python
d) Variable length positional arguments:
```

```python
Multiple Positional Arguments (*args)
*args is used in a function when you want to pass a variable
number of positional arguments.
```

```python
def add(*nums):
    return sum(nums)

print(add(10,30))
print(add(10, 20, 40, 50))
print(add())
```

```python
e) Variable length Keyword Arguments (**kwargs)
**kwargs allows us to pass any number of keyword arguments.
```

```python
def employee(**details):
    print(details.items())
    
employee(salary=40000, department="ECE", city="Bangalore")
employee(salary=80000, department="IT", Institute="MGIT")
employee(department="IT", Institute="MGIT", city="Hyderabad")
```

### 2. Scope of Variables

```python
a) Local variable → Declared inside a function, can only be
accessed within that function.
```

```python
def my_function():
    x = 10   # local variable
    print("Inside function, x =", x)

my_function()
# print(x)  # ❌ Error: x is local, not accessible outside
```

```python
b) Global variable → Declared outside any function, accessible
anywhere in the program.
```

```python
x = 50   # global variable

def my_function():
    print("Inside function, x =", x)

my_function()
print("Outside function, x =", x)
```

```python
c) Local Variable Shadows Global Variable.
```

```python
x = 20  # global variable

def my_function():
    x = 99   # local variable with same name
    print("Inside function, x =", x)

my_function()
print("Outside function, x =", x)
```
