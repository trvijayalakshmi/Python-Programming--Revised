### 2. Properties of Strings

```python
Properties of strings:
    
1) Immutable
2) Ordered & Indexed
3) Iterable
4) Supports slicing, concatenation, repetition
5) Case-sensitive
6) Works with membership operators
```

```python
Property 1. Immutable
```

```python
s = "hello"
# s[0] = 'H' ❌ Error Strings are immutable 
```

```python
Property 2. Ordered & Indexed - Each character has an index, 
starting from 0.
```

```python
word = "Python"
print(word[0])   # Ordered and Indexed
print(word[3])   
```

```python
Property 3. Strings are Iterable
```

```python
word = "Python"
for i in word:
    print(i)
```

```python
word = "PYTHON"
for i in range(len(word)):
    print(i)  # indices
```

```python
word = "PYTHON"
for i, j in enumerate(word):
    print(i, j)  # iterates with both index and value
```

```python
Property 4. Slicing - Extracting part of the string using [start:end]
(end index is excluded).

*** Explain string slicing with examples
```

```python
text = "Programming"
print(text[0:6])   # Progra
print(text[3:])    # gramming
print(text[:5])    # Progr
print(text[-3:-6:-1]) # Reverse printing
print(text[-8:-4])
print(text[::-1]) # Reverse of the string
```

```python
Property 4. Supports Concatenation and Repetition 
```

```python
a = "Hello"
b = "World"
print(a + b) # supports concatenation with +

print(3*a)  # supports repetition with *
```

```python
Property 5. Case sensitive
```

```python
a = "Hello"
b = "hello"
if a == b:
    print("Strings are same")
else:
    print("Strings are different")
```

```python
Property 6. Membership operators on strings
```

```python
a = "Hello"
print('e' in a)
print('s' in a)
```

### 3. String Methods - Built-in methods:

```python
Method 1. Case Conversion
```

```python
s = "python programming"
print(s.upper())   # PYTHON PROGRAMMING
print(s.lower())   # python programming
print(s.title())   # Python Programming
print(s.capitalize()) # Python programming

print(len(s))         # 18
```

```python
Method 2. Checking Content
```

```python
s = "Python123"
print(s.isalpha())   # False (contains numbers)
print(s.isdigit())   # False (contains letters)
print(s.isalnum())   # True (letters + numbers)
print(" ".isspace()) # True
print(s.isupper())
print(s.islower())
```

```python
Method 3. Searching and Replacing
```

```python
s = "I like Python and I also like Java"
print(s.find("like"))     # 2 (first index where found)
print(s.replace("Python", "Java"))  
print(s.replace(" ", "-"))
```

```python
Method 4. Splitting and Joining
```

```python
s = "I like Python and also Java"
words = s.split()    
print("Result of spliting", words)
print(type(words))

joined = "-".join(words)
print("Result of joining", joined)  
print(type(joined))
```

```python
Method 5. startswith, endswith and strip
```

```python
txt = "Hello, welcome to my world."

x = txt.startswith("Hello")
print(x)

y = txt.endswith(".")
print(y)

a = " Hello, World! "
print(a.strip())
print(a)
```

## Programs on Strings

```python
1. Check if a String is Palindrome
```

```python
text = input("Enter a string: ")

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
2. Count Words in a String and find the longest word in it.
```

```python
s = input("Enter a sentence: ")
words = s.split()
print("Number of words:", len(words))

longest = max(words, key=len)
print("Longest word:", longest)
```

```python
3. Count Vowels and Consonants
```

```python
s = input("Enter a string: ").lower()
vowels = "aeiou"
v_count = 0
c_count = 0

for ch in s:
    if ch.isalpha():
        if ch in vowels:
            v_count += 1
        else:
            c_count += 1

print("Vowels:", v_count)
print("Consonants:", c_count)
```
