```python
A tuple represents a sequence of objects separated by commas
and enclosed in parentheses. A tuple is an immutable object,
which means it cannot be changed, and we use it to represent
fixed collection of items.
```

### 1. Creating a Tuple

```python
# Empty tuple
t1 = ()
print(t1)   # ()

# Tuple with integers
t2 = (1, 2, 3, 4)
print(t2)   # (1, 2, 3, 4)

# Tuple with mixed data types
t3 = tuple(10, "apple", 3.14, True)
print(t3)   # (10, 'apple', 3.14, True)
```

```python
1. Single Element Tuple
```

```python
t5 = (10,)     # single element tuple
print(t5)      # (10,)
print(type(t5))

t6 = (10)      # just an integer
print(t6)      # 10
print(type(t6))
```

```python
2. Converting list and string to Tuples
```

```python
list_data = [1, 2, 3, 4]
t7 = tuple(list_data)
print(t7)  

string_data = "Python"
t8 = tuple(string_data)
print(t8) 
```

## ** 2. Explain any four Properties of tuples with examples.

```python
Properties of Tuples:
1. Ordered, Indexed, and Slicing.
2. Immutable → once created, elements cannot be changed.
3. Allows duplicates.
4. Heterogeneous: Can store different data types 
    (int, string, float, etc.).
5. Supports Concatenation and Repetition
5. Iterable.
6. Works with membership operators.
```

```python
Property 1. Tuples are Ordered, Indexed, and Slicing

Tuples maintain the order of elements as they are inserted.
```

```python
fruits = ("apple", "banana", "cherry", "mango", "grape")

print(fruits[1:4])   # From index 1 to 3 → ('banana', 'cherry', 'mango')
print(fruits[:3])    # From start to index 2 → ('apple', 'banana', 'cherry')
print(fruits[2:])
print(fruits[::-1])  # Reverses a tuple
```

```python
Property 2. Tuples are Immutable

Once created, the elements of a tuple cannot be changed
(no modification, addition, or deletion).
```

```python
t = (1, 2, 3)
t[0] = 5   # ❌ Error: Elements cannot be modified
print(t)      # Output: (1, 2, 3)
```

```python
Property 3. Tuples can contain Duplicate Elements

Unlike sets, tuples allow duplicate values.
```

```python
t = (1, 2, 2, 3, 1)
print(t)  # Output: (1, 2, 2, 3, 1)
```

```python
Property 4. Tuples can contain Heterogeneous Data

A tuple can store elements of different data types
(integers, strings, floats, lists, etc.).
```

```python
t = (10, "Python", 3.14, [1, 2, 3])
print(t)  
```

```python
Property 5. Supports Concatenation and Repetition
```

```python
t1 = ('a', 'b')
t2 = (3, 4)
print(t1 + t2)  # Concatenation
print(3*t1)
```

```python
Property 6. Iterable: Accessing Tuple Elements Using a Loop
```

```python
x = ("apple", "banana", "cherry")
for i in x:
    print(i)
```

```python
Looping through indicies
```

```python
x = ("apple", "banana", "cherry")
for i in range(len(x)):
    print(i)
```

```python
Looping through indicies and elements
```

```python
x = ("apple", "banana", "cherry") 
for i,j in enumerate(x):
    print(i, j)
```

```python
Property 7. Membership operator:
```

```python
x = ("apple", "banana", "cherry") 
print("banana" in x)
print("Python" in x)
```

### 3. Tuple Methods

```python
Since tuples are immutable, they have very few built-in methods:
Method a) count(x) → Returns number of times x appears.
Method b) index(x) → Returns index of first occurrence of x.
Method c) del - Deletes the tuple
```

```python
my_tuple = (10, 20, 30, 20, 40, 20)

print(my_tuple.count(20))   
print(my_tuple.index(20)) 
print("The length of tuple", len(my_tuple))
print("Max value", max(my_tuple))
print("Min value", min(my_tuple))
print(sorted(my_tuple))
```

```python
Method c) Deleting an element is not possible in a Tuple
Complete Tuple can be deleted
```

```python
t = ("apple", "banana", "cherry")
del t
print(t)
```

```python
Difference Between Tuples, Lists, and Sets in Python
```

```python
import sys
a_list = ['abc', 'xyz', 123, 231, 13.31, 0.1312]
a_tuple = ('abc', 'xyz', 123, 231, 13.31, 0.1312)
a_set = {'abc', 'xyz', 123, 231, 13.31, 0.1312}
print('The list size:', sys.getsizeof(a_list), 'bytes')
print('The tuple size:', sys.getsizeof(a_tuple), 'bytes')
print('The set size:', sys.getsizeof(a_set), 'bytes')
```

```python
# If Duplicates are there
import sys
a_list = ['abc', 'abc', 123, 123, 13.31, 13.31]
a_tuple = ('abc', 'abc', 123, 123, 13.31, 13.31)
a_set = {'abc', 'abc', 123, 123, 13.31, 13.31}
print('The list size:', sys.getsizeof(a_list), 'bytes')
print('The tuple size:', sys.getsizeof(a_tuple), 'bytes')
print('The set size:', sys.getsizeof(a_set), 'bytes')
```

### 4. Programs on Tuples

```python
1. Write a Python program to create a tuple with numbers 
and print all the elements at odd indices.
```

```python
numbers = (10, 20, 30, 40, 50, 60, 70, 80)

print("Tuple:", numbers)
print("Elements at odd indices:")

# Accessing elements at odd indices using slicing
for i in range(1, len(numbers), 2):   # start from index 1, step = 2
    print(numbers[i])
```
