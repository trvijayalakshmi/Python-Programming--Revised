```python
A dictionary is a collection of key–value pairs.
```

### 1. Creating Dictionaries

```python
student1 = {"id": 1, "name": "Ravi", "marks": 90}
student2 = dict(id = 1, name = "Ravi", marks = 90)
student3 = dict([("id", 1), ("name", "Ravi"), ("marks", 90)])
student4 = {}  # empty dictionary

print(student1)
print(student2)
print(student3)
print(student4)
print(len(student1))
print(len(student4))
```

### Properties of Dictionary

```python
Properties of Dictionary:
1) Ordered (Unless modified)
2) Keys are unique & immutable
3) Values are Mutable
4) Keys and Values can be of any datatype 
5) Accessed and indexed by keys
6) Supports Membership Operators for Keys
7) Dynamic and iterable
```

```python
Property 1. Ordered (Unless modified)
```

```python
student = {"id": 1, "name": "Ravi", "marks": 90}
print(student)
student = {"id": 1, "name": "Ravi", "marks": 90, "id": 23}
print(student)
```

```python
Property 2. Keys are unique and immutable, but values can be duplicated.
```

```python
student = {"id": 23, "name": "Ravi", "marks": 23, "name": "Gopi"}
print(student)
```

```python
# Change a single key
student = {"id": 23, "name": "Ravi", "marks": 23, "name": "Gopi"}

student['totalmarks'] = student.pop('marks')
print(student)
```

```python
Property 3. Values are Mutable (modifying values)
```

```python
# Changing values and adding new items
student["name"] = "Sanjay"   # changing values in dict
student["totalmarks"] = 95   # changing values in dict
student['age'] = 23   # adding new items to dict
print(student) 
```

```python
Property 4. Keys and values can be of any datatype
```

```python
student2 = {1:"a", "b":True, "c":3.14}
print(student2)
```

```python
Property 5. Accessed and indexed by keys.
```

```python
student = {'id': 23, 'name': 'Ravi', 'marks': 90}
print(student["name"])      
print(student.get("age"))
```

```python
Property 6. Dictionary Membership Test: Checks only Keys and not Values
```

```python
student = {'id': 23, 'name': 'Ravi', 'marks': 90}
print('name' in student) 
print('marks' not in student) 
print(23 in student) 
```

### 3. Dictionary Methods

```python
Method 1. To access key-values from dict
keys, values, and items of the dictionary can be extracted
```

```python
student = {"id": 1, "name": "Ravi", "marks": 90}

print(student.keys())     # dict_keys(['id', 'name', 'marks'])
print(student.values())   # dict_values([1, 'Ravi', 90])
print(student.items())    # dict_items([('id', 1), ('name', 'Ravi'), ('marks', 90)])
```

```python
Property 7. Dictionaries can be iterated:
```

```python
a) Iterating over keys
```

```python
person = {"name": "Rahul", "age": 22, "city": "Delhi"}
for key in person.keys():
    print(key)
```

```python
b) Iterating over values
```

```python
person = {"name": "Rahul", "age": 22, "city": "Delhi"}
for value in person.values():
    print(value)
```

```python
c) Iterating over key-value pairs
```

```python
person = {"name": "Rahul", "age": 22, "city": "Delhi"}
for i, j in person.items():
    print(i, ":", j)
```

```python
Method 2. Updating the dictionary
```

```python
student.update({"marks": 95, "grade": "A"})
print(student)
```

```python
Method 3. Get the specified item from the dictionary
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}
print(student.get("marks"))
print(student["marks"])
```

```python
Using a default value for get() method
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}
print(student.get("age",'Not found'))   
print(student.get("gender"))
print(student.get("gender", "Not Found"))  # Default value if key not found
```

```python
Difference between .get() vs [] 
If key is not found .get() gives None 
However if key is not found using [] gives Keyerror
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}
print(student.get('gender'))  # ✅ None
print(student['gender'])   # ❌ KeyError
```

```python
Method 4. Removing item from the dictionary (Key should be provided)
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}
student.pop("id")      # removes 'id'
print(student)
```

```python
Method 5. Pop the last item (key need not be provided)
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}
student.popitem()      # removes last one
print(student)
```

```python
Method 6. Clearing the dictionary
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}
student.clear()        # clears dictionary
print(student)
```

```python
Method 7. Deleting an item from dictionary and deleting the entire 
dictionary
```

```python
student = {'id': 1, 'name': 'Ravi', 'marks': 95, 'grade': 'A'}

del student["name"]  # deleting an item from dict
print(student)

del student   #Deletes the entire dict
print(student)
```

### 4. Dictionary Comprehension

```python
Like list comprehension, dict comprehension can be created with 
key: value pairs.
```

```python
1. Squares of numbers using dictionary comprehension
```

```python
# with comprehension
squares = {x: x*x for x in range(1, 11)}
print(squares)   # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```

```python
# without comprehension
squares = {}
for x in range(1,11):
    squares[x] = x*x
print(squares)
```

```python
2. Squares of even numbers using dictionary comprehension
```

```python
squares = {x: x*x for x in range(1, 10) if x%2==0}
print(squares) 
```

```python
3. Create a Dictionary of Odd/Even Labels for the numbers over
the range 1 to 15
```

```python
labels = {n: ("Even" if n % 2 == 0 else "Odd") for n in range(1, 16)}
print(labels)
```

```python
4. Write a Python program to count the frequency of each character
in a given string using a dictionary.
```

```python
# With comprehension
text = "banana"
freq = {ch: text.count(ch) for ch in text}
print(freq)  
```

```python
# Without comprehension
text = "banana"

freq = {}
for ch in text:
    freq[ch] = text.count(ch)

print("Character Count:", freq)
```

```python
5. Write a Python program to count the frequency of each word
in a given string using a dictionary.
```

```python
text = "apple banana apple orange banana apple"
words = text.split()

freq = {}
for i in words:
    freq[i] = words.count(i)

print("Word Frequency:", freq)
```

```python
6. Inverting a Dictionary
```

```python
# With comprehension
original = {"a": 1, "b": 2, "c": 3}
inverted = {value: key for key, value in original.items()}
print(inverted)
```

```python
# Without Comprehension
my_dict = {"a": 1, "b": 2, "c": 3}
print("Original Dict", my_dict)

inverted_dict = {}
for key, value in my_dict.items():
    inverted_dict[value] = key
    
print("Inverted Dict", inverted_dict)
```

```python
7. Remove Duplicate Values in a Dictionary
```

```python
data = {"a": 104, "b": 130, "a": 105, "d": 130}
print("Before Removing Duplicates:", data)
data2 = {}
for key, value in data.items():
    if value not in data2.values():
        data2[key] = value

print("After Removing Duplicates:", data2)
```

```python
8. Find the Student with Maximum marks
```

```python
x = {"Alice": 85, "Bob": 92, "Charlie": 89}
#For single topper
highest = max(x.values())
topper = max(x, key=x.get)
print(f"Topper is: {topper} with {highest} marks")
```

```python
To find all the toppers
```

```python
x = {"Alice": 85, "Bob": 92, "Charlie": 92}
#For multiple toppers
highest = max(x.values())
toppers = []
for key, value in x.items():
    if value == highest:
        toppers.append(key)

print(f"Toppers are: {toppers} with {highest} marks")
```

```python
# For multiple toppers Using list comprehension
x = {"Alice": 85, "Bob": 92, "Charlie": 92}
highest = max(x.values())
toppers = [key for key, value in x.items() if value == highest]
print(f"Toppers are: {toppers} with {highest} marks")
```

```python
9. Program to store names and marks of 'n' students in a 
dictionary and print the toppers with maximum marks.
```

```python
students = {}
x = {"Alice": 85, "Bob": 92, "Charlie": 92}

n = int(input('Enter the number of students'))
for i in range(n):
    name = input(f"Enter name of student {i+1}: ")
    marks = int(input(f"Enter marks of {name}: "))
    students[name] = marks

# Find the student with the highest marks
highest = max(students.values())

# Find all toppers using list comprehension   
toppers = [name for name, marks in students.items() if marks == highest]

print("\nStudents and their marks:", students)
print(f"Toppers: {toppers} with {highest} marks")
```
