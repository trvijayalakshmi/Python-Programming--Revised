# Object Oriented Programming

### Theory for Classes and Objects:

```python
a) Class: A class is a blueprint for creating objects. It defines attributes 
   (variables) and methods (functions).
b) Attributes: Attributes are variables inside a class. They describe properties of
   an object.
c) Methods: Methods are functions inside a class. They define
   behavior of objects.
d) Object: Every object will have attributes and methods.
```

### Programs on Classes and Objects

```python
1. Define a class "Calculator" with appropriate methods.
   Instantiate an object and print the results.
```

```python
class Calculator:
    def add(self, a, b): return a + b
    def sub(self, a, b): return a - b
    def mul(self, a, b): return a * b
    def div(self, a, b): return a / b 

c = Calculator()
print(c.add(10, 5))
print(c.sub(10, 5))
print(c.mul(10, 5))
print(c.div(10, 4))
```

```python
2. Define a class with appropriate attributes and methods to
   check whether a student has cleared exam or not.
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    def result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

s1 = Student("Anu", 35)
s1.display()
s1.result()
```

```python
3. Define a class with appropriate attributes and methods for 
   banking application.
```

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance  

    def deposit(self, amount):
        self.balance += amount
    
    def withdraw(self, amount):
        self.balance -= amount

    def show_balance(self):
        print("Balance:", self.balance)
    
    def check_balance(self):
        if self.balance < 1000:
            print("Minimum balance not maintained")
        else:
            print("Balance is sufficient")

b = Bank(1000)
b.deposit(500)
b.withdraw(200)
b.show_balance()
b.check_balance()
```

### Theory for Inheritance

```python
Inheritance allows a class (child) to inherit properties and methods
from another class (parent).

Uses of Inheritance:
a) Since a child class can inherit all the functionalities of the
parent's class, this allows code reusability.
b) Once a functionality is developed, you can simply inherit it. 
No need to reinvent the wheel.
c) Since you can also add your own functionalities in the child 
class, you can inherit only the useful functionalities and define 
other required features.

Types of Inheritance:
    1. Single inheritance
    2. Multiple Inheritance
    3. Multilevel Inheritance
    4. Hierarchical Inheritance
```

### Programs on Inheritance

```python
4. Write a program using Multilevel inheritance (chain of inheritance)
    Person → Employee → Manager.
Implement multilevel inheritance with a simple example.
```

```python
class Person:
    def basic_info(self):
        print("Learning things to find a job in a company")

class Employee(Person):
    def role(self):
        print("Attend works assigned by the company")

class Manager(Employee):
    def decision(self):
        print("Manager makes decisions")

# Create object
s = Manager()

s.basic_info()   # method from Person
s.role()       # method from Employee
s.decision()     # method from Manager or own method
```

```python
5. Define a child class that inherits methods from parent classes.
   i.e. Multiple inheritance (Teacher + Researcher = Professor).
    
Write a program to demonstrate Multiple Inheritance.
```

```python
class Teacher:
    def teach(self):
        print("Teaches the subject Python")

class Researcher:
    def research(self):
        print("Does research in Deep learning")
    
class Professor(Teacher, Researcher):   # Inherites both the features
    def display(self):
        print("Teaches the subjects and does research")

p = Professor()
p.teach()
p.research()
p.display()
```

```python
6. Write a program that demonstrates Hierarchical inheritance.
```

```python
# Parent class
class Employee:
    def display_company(self):
        print("Company: TCS")

# Child class 1
class Manager(Employee):
    def show_role(self):
        print("Role: Manager")

# Child class 2
class Developer(Employee):
    def show_role(self):
        print("Role: Developer")

# Creating objects
m = Manager()
d = Developer()

# Accessing methods
m.display_company()
m.show_role()

d.display_company()
d.show_role()
```

### Theory for Polymorphism

```python
Polymorphism allows different classes to define methods with the same name
but different behaviors. Same method name behaves differently 
depending on context.

Uses of Polymorphism:
a) Improves code flexibility and reusability
b) Makes code easier to maintain and extend
c) Reduces duplication of logic

Types of Polymorphism:
a) Method Overriding: A subclass provides its own implementation 
   of a method already defined in its parent class.
b) Method Overloading: Same method name with different parameters.
```

### Programs on Polymorphism

```python
7. Write a program to demonstrate polymorphism using 
   "Method Overriding".
A) Python achieves polymorphism by using super() method
 to overcome “method overriding” 
```

```python
class Employee:   # Parent class
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Employee Name:", self.name)

class Manager(Employee):   # Inheriting from Employee (Child class)
    def display(self):
        Employee.display(self)
#         super().display()
        print("Role: Manager")

# Create object of child class
m = Manager("Ravi")

m.display()     # calling overrided method
```

```python
8. Write a program to demonstrate polymorphism using 
   "Method Overloading with different parameters". 

Discuss how Python achieves polymorphism without method overloading. 
A) Python achieves polymorphism without "Method Overloading" by 
using default arguments or variable number of arguments (*args).
```

```python
class Calculator:
    def add(self, a=0, b=0, c=0):
        return a + b + c

c = Calculator()
print(c.add(2, 3))       
print(c.add(2, 3, 4))    
print(c.add(5))          
```

```python
class Calculator:
    def add(self, *args):
        return sum(args)

c = Calculator()
print(c.add(2, 3))       # 5
print(c.add(2, 3, 4))    # 9
print(c.add(5)) 
```
