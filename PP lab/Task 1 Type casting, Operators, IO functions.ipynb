## 1. Type Casting

### Theory for Type Casting:

```python
Type Casting: Converting from one data type to another.

Two types of Type casting: 
• Implicit Type Casting → Done automatically by Python.
• Explicit Type Casting → Done manually by the programmer using built-in functions.
```

### Programs on Type Casting

#### a. Implicit Type Casting

```python
# Eg1: Integer to float
x = 10      # int
y = 3.5    # float
result = x + y  # int + float → float
print(result)   
print(type(result))

# Eg2: Integer to Complex
x = 4    # int
y = 2 + 3j  # complex
result = x + y
print(result)        # (6+3j)
print(type(result))  # <class 'complex'>
```

#### b. Explicit Type casting

```python
# Eg1: Converting floating point to integer
x = int(3.14)
print(x) 
print(type(x))

# Eg2: Converting string to floating point
x = float("3.14")
print(x) 
print(type(x))

# Eg3: Converting integer to string
x = str(123)
print(x)
print(type(x))
```

## 2. Operators

### Theory for Operators:

```python
1. Arithmetic Operators: +, -, *, /, //, %, ** 
2. Relational Operators: >, <, >=, <=, ==, != 
3. Logical Operators: and, or, not
4. Assignment Operators: +=, -=, *=, /=, =, //=
5. Membership Operators: in, not in
6. Identity Operators: is, is not
7. Bitwise Operators: |, &, ^, <<, >>
```

### Programs on Operators

```python
# 1. Arithmetic Operators

x = 10
y = 3
result1 = x + y * 2 
result2 = x // y
print(result1)
print(result2)
```

```python
# 2. Relational Operators

a, b = 10, 20
print("a > b :", a > b)     # False
print("a < b :", a < b)     # True

x, y = 15, 15
print("x == y :", x == y)   # True
print("x != y :", x != y)   # False
```

```python
# 3. Logical operators
x, y = True, False
print("x and y:", x and y)   
print("x or y:", x or y)
print("not x:", not x)
```

```python
# 4. Assignment operators

x = 20
x /= 5  
print("Result:", x)

x = 17
x %= 5 
print("Result:", x)
```

```python
# 5. Membership operators

x = [1, 2, 3, 4, 5]
print(3 in x)  # Output: True
print(6 not in x)  # Output: True
```

```python
# 6. Identity operators

x = [1, 2, 3]
y = [1, 2, 3]

print(x == y)   # True → contents are the same
print(x is y)   # False → stored at different memory locations
print(x is not y)
```

```python
# 7. Bit-wise operators

a = 12  # 1100
b = 6   # 0110

print("a & b =", a & b)   # 0100 → 4        (a and b)
print("a | b =", a | b)   # 1110 → 14     (a or b)
print("a ^ b =", a ^ b)   # 1010 → 10     (a xor b)
print("a << 2 =", a << 2) # 110000 → 48   (Left shift)
print("a >> 2 =", a >> 2) # 0011 → 3    (Right shift)
```

### 3. Input and Print functions

### Theory for Input and Print functions:

```python
input(): The default datatype of input() function is str
print(): It helps to show messages, variables, and results to the user
```

### Programs using Input and Print functions

```python
1. Write a program to calculate compound interest when principal,
rate and number of periods are given.
```

```python
principal = float(input("Enter the principal amount: "))
rate = float(input("Enter the annual interest rate (in percentage): "))
time = int(input("Enter the number of periods: "))

amount = principal * (1 + rate/100) ** time
interest = amount - principal

# Different ways of using print function
print("The compound interest is:", interest)
print("The compound interest is:", format(interest, ".2f")) 
print(f"The compound interest is: {interest:.2f}")
```

```python
2. Given coordinates (x1, y1), (x2, y2) find the distance between 
two points.
```

```python
x1 = float(input("Enter the x-coordinate of the first point: "))
y1 = float(input("Enter the y-coordinate of the first point: "))
x2 = float(input("Enter the x-coordinate of the second point: "))
y2 = float(input("Enter the y-coordinate of the second point: "))

distance = (((x2 - x1)**2 + (y2 - y1)**2)**.5)
print(f"The distance between the two points is: {distance:.2f}")
```
