### Theory for Loops:

```python
Loops help in automation and reducing code length.
They execute a set of statements repeatedly based on a condition.
Useful when working with lists, strings, tuples and dictionary.
```

### Programs using Loops:

```python
1. Program to find factorial of a number  using for and while loops
```

```python
num = int(input("Enter a number: "))
fact = 1
for i in range(1, num+1):
    fact *= i
print("Factorial is:", fact)
```

```python
num = int(input("Enter a number: "))
fact = 1
i = 1
while i <= num:
    fact *= i
    i += 1
print("Factorial is:", fact)
```

```python
2. Python program to print the Fibonacci sequence using for and while loops
```

```python
n = int(input("Enter n: "))
a, b = 0, 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a+b
```

```python
n = int(input("Enter n: "))
a, b = 0, 1
i = 0
while i < n:
    print(a, end=" ")
    a, b = b, a + b
    i += 1
```

```python
3. Python program to find the gcd of the given numbers
```

```python
a = 25
b = 10
while b:
    a, b = b, a % b
print(f"The gcd is {a}")
```

```python
4. Multiplication Table of a Number
```

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num*i}")
```

```python
5. Python program to print all prime numbers in an interval

Eg. program using break
```

```python
s = 6
e = 30
for i in range(s, e):
        if i == 0 or i == 1:
            continue
        else:
            for j in range(2, i//2+1):
                if i % j == 0:
                    break
            else:
                print(i,end = " ")
```

```python
6. Program to check whether a given no is prime or not
```

```python
i = int(input("Enter a number: "))

if i > 1:
    for j in range(2, i//2+1):
        if i % j == 0:
            print(i, "is not a prime number")
            break
    else:
        print(i, "is a prime number")
else:
    print(i, "is not a prime number")
```

```python
7. Reverse the digits of a number
```

```python
num = int(input("Enter a number: "))
rev = 0
while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10
print("Reversed number is:", rev)
```

```python
p = 69
for i in range(5, 0, -1):  
    for j in range(i):   
        print(chr(p), end=" ") 
    print()
    p -= 1
```
