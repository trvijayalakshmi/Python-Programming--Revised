### a) Dictionary is a Mapping Type

### Theory for Dictionary:

```python
A dictionary is a collection of key–value pairs.

Properties of Dictionary:
    
1) Mutable
2) Keys are unique & immutable
3) Values can be any type 
4) Ordered (Unless modified)
5) Accessed and indexed by keys
6) Dynamic and iterable
7) Supports Membership Operators
```

### Programs using Dictionary:

```python
1. Program to find the Student with Maximum Value
```

```python
x = {"Alice": 85, "Bob": 92, "Charlie": 78}
highest = max(x.values())
topper = max(x, key=x.get)
print(f"Topper is: {topper} with {highest} marks")
```

```python
#For multiple toppers
x = {"Alice": 85, "Bob": 92, "Charlie": 92}

highest = max(x.values())
toppers = []
for key, value in x.items():
    if value == highest:
        toppers.append(key)

print(f"Toppers are: {toppers} with {highest} marks")
```

```python
2. Create a Dictionary of Cubes of even numbers using dictionary
comprehension.
```

```python
cubes = {i: i**3 for i in range(1, 11) if i%2==0}
print(cubes) 
```

```python
# without comprehension
cubes = {}
for i in range(1,11):
    if i%2 == 0:
        cubes[i] = i**3
print(cubes)
```

```python
3. Inverting a Dictionary
```

```python
# Without Comprehension
data = {"a": 40, "b": 20, "c": 10, "d": 30}

print("Original Dict", data)
inverted_dict = {}
for key, value in data.items():
    inverted_dict[value] = key
    
print("Inverted Dict", inverted_dict)
```

```python
# with dictionary comprehension
data = {"a": 40, "b": 20, "c": 10, "d": 30}

inverted = {value: key for key, value in data.items()}
print(inverted)
```

```python
4. Remove Duplicate Values from a Dictionary
```

```python
data = {"a": 40, "a": 20, "c": 10, "d": 10}
print("Before Removing Duplicates:", data)

unique_data = {}
for key, value in data.items():
    if value not in unique_data.values():
        unique_data[key] = value

print("After Removing Duplicates:", unique_data)
```

```python
5. Write a Python program to create a dictionary of characters
and their frequency in a given string using dictionary comprehension.
```

```python
s = "programming"

freq = {}
for ch in s:
    freq[ch] = s.count(ch)

print("Character Count:", freq)
```

```python
s = "programming"
freq = {ch: s.count(ch) for ch in s}
print("Character Count:", freq)
```

```python
6. Write a Python program to create a dictionary of words and their
frequency in a given sentence using dictionary comprehension.
```

```python
# Using Dictionary comprehension
s = "Python is easy and Python is powerful"
words = s.split()
freq = {i: words.count(i) for i in words}
print("Character Count:", freq)
```

```python
s = "Python is easy and Python is powerful"
words = s.split()

freq = {}
for i in words:
    freq[i] = words.count(i)

print("Character Count:", freq)
```

### b) Theory for Sets

```python
Properties of Sets:
1) Unordered
2) Mutable
3) Unique (no duplicates)
4) Heterogeneous
5) No indexing or slicing
6) Efficient for mathematical and membership operations
```

### Programs on Sets

```python
7. Program to count unique words in a sentence
```

```python
sentence = "Python is easy and Python is powerful"
words = sentence.split()
unique_words = set(words)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```

```python
8. Program to count unique characters in a string
```

```python
text = "programming"
print("Unique characters:", set(text))
unique_chars = "".join(set(text))
print("Unique characters:", unique_chars)
```

```python
9. Program to find common elements and elements only in the first set
```

```python
# Create two sets
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Finding common elements
common = set1.intersection(set2)
# Finding elements present only in the first set
only_in_first = set1.difference(set2)

# Displaying results
print("Common Elements:", common)
print("Elements only in Set 1:", only_in_first)
```
